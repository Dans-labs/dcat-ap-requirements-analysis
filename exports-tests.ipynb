```python
import pyshacl

def validate_shacl(data_graph, shacl_graph, inference='rdfs', abort_on_first=False, meta_shacl=False, debug=False):
    """
    Validates a data graph against a SHACL graph.

    :param data_graph: The data graph to validate.
    :param shacl_graph: The SHACL graph containing the shapes.
    :param inference: The type of inference to apply (default is 'rdfs').
    :param abort_on_first: Whether to abort on the first validation error (default is False).
    :param meta_shacl: Whether to validate the SHACL graph itself (default is False).
    :param debug: Whether to enable debug mode (default is False).
    :return: A tuple containing the validation result, a validation report graph, and a boolean indicating conformity.
    """
    try:
        conforms, report_graph, report_text = pyshacl.validate(
            data_graph,
            shacl_graph=shacl_graph,
            inference=inference,
            abort_on_first=abort_on_first,
            meta_shacl=meta_shacl,
            debug=debug
        )
        return conforms, report_graph, report_text
    except Exception as e:
        print(f"An error occurred during SHACL validation: {e}")
        return False, None, str(e)
    

```

```python
# testing https://portal.staging.odissei.nl/api/datasets/export?exporter=dcat3-turtle&persistentId=doi%3A10.34894/WXKJHY
conforms, report_graph, report_text =validate_shacl(data_graph="example-datasets/8575OP.ttl", shacl_graph="SHACL/DANS-shacl.ttl", inference='rdfs', abort_on_first=False, meta_shacl=False, debug=False)
print(f"Conforms: {conforms}\nReport Graph: {report_graph}\nReport Text: {report_text}")
```
