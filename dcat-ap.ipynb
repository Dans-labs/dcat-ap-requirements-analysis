```python
# parse DCAT as RDF graph
import re
import rdflib
from rdflib.plugins.sparql import prepareQuery
from SPARQLWrapper import SPARQLWrapper, TURTLE, CSV, JSON

prefixes = '''
PREFIX adms: <http://www.w3.org/ns/adms#>
PREFIX cc: <http://creativecommons.org/ns#>
PREFIX dcat: <http://www.w3.org/ns/dcat#>
PREFIX dcat: <http://www.w3.org/ns/dcat#>
PREFIX dcatap: <http://data.europa.eu/r5r/>
PREFIX dcatap_shapes: <https://semiceu.github.io/DCAT-AP/releases/3.0.0/html/shacl/shapes.ttl#>
PREFIX dcatap_ranges: <https://semiceu.github.io/DCAT-AP/releases/3.0.0/html/shacl/range.ttl#>
PREFIX dct: <http://purl.org/dc/terms/>
PREFIX foaf: <http://xmlns.com/foaf/0.1/>
PREFIX lcon: <http://www.w3.org/ns/locn#>
PREFIX org: <http://www.w3.org/ns/org#>
PREFIX owl: <http://www.w3.org/2002/07/owl#>
PREFIX odrl: <http://www.w3.org/ns/odrl/2/>
PREFIX prov: <http://www.w3.org/ns/prov#>
PREFIX rdf: <http://www.w3.org/1999/02/22-rdf-syntax-ns#>
PREFIX rdfs: <http://www.w3.org/2000/01/rdf-schema#>
PREFIX schema: <http://schema.org/>
PREFIX sh: <http://www.w3.org/ns/shacl#>
PREFIX skos: <http://www.w3.org/2004/02/skos/core#>
PREFIX spdx: <http://spdx.org/rdf/terms#>
PREFIX time: <http://www.w3.org/2006/time#>
PREFIX vcard: <http://www.w3.org/2006/vcard/ns#>
PREFIX xsd: <http://www.w3.org/2001/XMLSchema#>
'''

def prefixes2dict(prefixes_str):
    prefixes_dict = {}
    for line in prefixes_str.split('\n'):
        found = re.findall(r'PREFIX\W(?P<prefix>.*):\W{1,}<(?P<ns>.*)>', line)
        if found != []:
            prefix, ns = found[0]
            prefixes_dict[prefix] = ns
    return prefixes_dict


def bind_prefixes2graph(prefixes_str, graph):
    for line in prefixes_str.split('\n'):
        found = re.findall(r'PREFIX\W(?P<prefix>.*):\W{1,}<(?P<ns>.*)>', line)
        if found != []:
            prefix, ns = found[0]
            graph.bind(prefix, ns)
    return graph

dcat_g = rdflib.Graph()
dcat_g=bind_prefixes2graph(prefixes_str=prefixes, graph=dcat_g)
dcat_g.parse("http://www.w3.org/ns/dcat#")

```

```python
# labels and definitions of classes:     dcat:DataService, dcat:Dataset, dcat:Distribution

q_dcat_Dataset =  prefixes + '''
CONSTRUCT{
    ?subj a ?class ; 
          rdfs:label ?label ;
          skos:definition ?def ;
          skos:scopeNote ?scope ;           
}
WHERE {
  VALUES ?subj {
    dcat:Catalog
    dcat:DataService
    dcat:Dataset
    dcat:Distribution
      }
    ?subj a ?class ;
          skos:definition ?def ;
          rdfs:label ?label ; 
    FILTER( LANG(?def) = 'en'&&  LANG(?label) = 'en')         
} 
'''
```

```python
qr = dcat_g.query(q_dcat_Dataset)
construct_graph = qr.graph
print(str(construct_graph.serialize()))
```

# DCAT-AP 3.0.0

Maintained by the [SEMIC Support Centre](https://interoperable-europe.ec.europa.eu/collection/semic-support-centre). 

> DCAT-AP is a DCAT profile for sharing information about Catalogues containing Datasets and Data Services descriptions in Europe, under maintenance by the SEMIC action, Interoperable Europe.

> This Application Profile provides a minimal common basis within Europe to share Datasets and Data Services cross-border and cross-domain. 
https://semiceu.github.io/DCAT-AP/releases/3.0.0/




> DCAT-AP allows:
> * Data catalogues to describe their dataset collections using a standardised description, while keeping their own system for documenting and storing them.
> * Content aggregators, such as the European Data Portal, to aggregate such descriptions into a single point of access.
>  * Data consumers to more easily find datasets through a single point of access.

https://interoperable-europe.ec.europa.eu/collection/semic-support-centre/solution/dcat-application-profile-data-portals-europe

DCAT-AP purpose of facilitatting the reuse of data, in concrete: 

> * It proposes mandatory, recommended or optional classes and properties to be used for a particular application; 
> * It identifies requirements to control vocabularies for this particular application; 
> * It gathers other elements to be considered as priorities or requirements for an application such as a conformance statement.
    
https://interoperable-europe.ec.europa.eu/collection/semic-support-centre/solution/dcat-application-profile-data-portals-europe


```python
# mandatory properties for dcat:Calog dcat:DataService dcat:Dataset

dcatap_shapes_g = rdflib.Graph()
dcatap_shapes_g=bind_prefixes2graph(prefixes_str=prefixes, graph=dcatap_shapes_g)

dcatap_shapes_g.parse("https://semiceu.github.io/DCAT-AP/releases/3.0.0/html/shacl/shapes.ttl")
dcatap_shapes_g.parse("https://semiceu.github.io/DCAT-AP/releases/3.0.0/html/shacl/range.ttl")
dcatap_shapes_g.parse("https://semiceu.github.io/DCAT-AP/releases/3.0.0/html/shacl/shapes_recommended.ttl")


dcatap_mandatory =  prefixes + '''
SELECT ?class_ ?prop ?req ?type
FROM <https://semiceu.github.io/DCAT-AP/releases/3.0.0/html/shacl/shapes.ttl> 
FROM NAMED <https://semiceu.github.io/DCAT-AP/releases/3.0.0/html/shacl/range.ttl> 
FROM NAMED <https://semiceu.github.io/DCAT-AP/releases/3.0.0/html/shacl/shapes_recommended.ttl> 
WHERE {
   {


      GRAPH <https://semiceu.github.io/DCAT-AP/releases/3.0.0/html/shacl/shapes.ttl> {
      VALUES ?class_ {
        dcat:Catalog
        dcat:DataService
        # dcat:Dataset
      }
      ?shape sh:targetClass ?class_ ;
             sh:property ?b_node . 

        # mandatory prop: sh:severity sh:Violation ; sh:minCount 1
        ?b_node sh:path ?prop ;
                sh:nodeKind ?type ;
                sh:severity sh:Violation ;
                sh:minCount 1  . 
        BIND("mandatory" AS ?req)     
      }

#       GRAPH <https://semiceu.github.io/DCAT-AP/releases/3.0.0/html/shacl/shapes_recommended.ttl> { 
#         #recommend prop: sh:severity sh:Warning ; sh:minCount 1 
#         ?b_node sh:path ?prop ;
#                 sh:nodeKind ?type ;
#                 sh:severity sh:Warning ;  # this is in another graph(file)
#                 #       sh:minCount 1  .      
#         BIND("recommended" AS ?req)     
#       }      
   }
}
ORDER BY ?class_ ?req
'''
# qp = prepareQuery(
#     dcatap_mandatory,
#     initNs = prefixes2dict(prefixes)
# )

qr = dcatap_shapes_g.query(dcatap_mandatory, 
                           # initBindings=prefixes2dict(prefixes)
                           )
# print(qr.serialize(format=JSON))
for row in qr:
   
   print(f"{row.class_} - property {row.prop} requirement:{row.req} type: {row.type}")



# TODO: display prefixes - despite adding the prefix to initBindings & initNs
```
